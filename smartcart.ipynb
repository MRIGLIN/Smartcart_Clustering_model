```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("smartcart_customers.csv")
```

```python
df.isnull().sum()
```

```python
#data preprocessing
```

```python
#handling missing values
```

```python
df["Income"] =  df["Income"].fillna(df["Income"].median())
```

```python
df
```

```python
df.isnull().sum()
```

```python
#feature engineering
```

```python
df["Age"] = 2026 -  df["Year_Birth"]
```

```python
df
```

```python
df["Dt_Customer"] = pd.to_datetime(df["Dt_Customer"],dayfirst = True)
ref_date = df["Dt_Customer"].max()
df["With us since"] = (ref_date - df["Dt_Customer"]).dt.days
```

```python
df
```

```python
#spending
```

```python
df.columns
```

```python
df["Total spending"] = df["MntWines"] + df["MntFruits"] + df["MntMeatProducts"] + df["MntFishProducts"] + df["MntSweetProducts"] + df["MntGoldProds"]
```

```python
df.head()
```

```python
#children
```

```python
df["Children"] = df["Kidhome"] + df["Teenhome"]
```

```python
df.head()
```

```python
#education
```

```python
df["Education"].value_counts()
```

```python
df["Education"] = df["Education"].replace(
    {"Graduation":"Grad","PhD":"PG","Master":"PG","2n Cycle" : "UG","Basic":"UG"}
)
```

```python
df["Education"].value_counts()
```

```python
#marital status
```

```python
df["Lives With"] = df["Marital_Status"].replace(
    {"Married":"partner","Together":"partner","Single":"Self","Divorced":"Self","Widow":"Self","Alone":"Self","Absurd":"Self","YOLO":"Self"}
)
```

```python
df
```

```python
df["Lives With"].value_counts()
```

```python
#drop columns 
```

```python
df.columns
```

```python
clean_df = df.drop(columns = ["ID","Year_Birth", "Marital_Status","Kidhome", "Teenhome","Dt_Customer", "MntWines","MntFruits","MntMeatProducts","MntFishProducts","MntSweetProducts","MntGoldProds"])
```

```python
clean_df
```

```python
#encoding
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe = OneHotEncoder()
catagorical_cols = ["Education","Lives With"]
```

```python
encoded_cols = ohe.fit_transform(clean_df[catagorical_cols])
```

```python
encoded_df = pd.DataFrame(encoded_cols.toarray(),columns = ohe.get_feature_names_out(catagorical_cols),index = clean_df.index)
```

```python
encoded_df
```

```python
final_encoded_df = pd.concat([clean_df.drop(columns = catagorical_cols),encoded_df],axis = 1)
```

```python
final_encoded_df
```

```python
final_encoded_df.shape
```

```python
#scaling
```

```python
from sklearn.preprocessing import StandardScaler 
```

```python
x = final_encoded_df
```

```python
scaler = StandardScaler()
```

```python
x_scaled = scaler.fit_transform(x)
```

```python
#visualing
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components = 3)
```

```python
x_pca = pca.fit_transform(x_scaled)
```

```python
fig = plt.figure(figsize = (8,6))
ax = fig.add_subplot(111,projection = "3d")
ax.scatter(
    x_pca[:,0],x_pca[:,1],x_pca[:,2]
)
ax.set_xlabel("PCA1")
ax.set_ylabel("PCA2")
ax.set_zlabel("PCA3")
ax.set_title("3d projection")
```

```python
from sklearn.cluster import KMeans
```

```python
from kneed import KneeLocator
```

```python
wcss = []
for k in range(1,11):
    kmeans = KMeans(n_clusters = k,random_state = 42)
    kmeans.fit_predict(x_pca)
    wcss.append(kmeans.inertia_)
```

```python
knee = KneeLocator(range(1,11),wcss,curve = "convex",direction = "decreasing")
optimal_k = knee.elbow
```

```python
print("best k =",optimal_k)
```

```python
plt.plot(range(1,11),wcss,marker = "o")
plt.xlabel("k")
plt.ylabel("wcss")
```

```python
    #silhouette score
```

```python
from sklearn.metrics import silhouette_score
```

```python
scores = []
for k in range(2,11):
    kmeans = KMeans(n_clusters = k,random_state = 42)
    labels = kmeans.fit_predict(x_pca)
    score = silhouette_score(x_pca,labels)
    scores.append(score)
```

```python
plt.plot(range(2,11),scores,marker = "o")
plt.xlabel("k")
plt.ylabel("silhouette score")
```

```python
k_range = range(2,11)
fig,ax1 = plt.subplots(figsize = (8,6))
ax1.plot(k_range,wcss[:len(k_range)],marker = "o",color = "green")
ax1.set_xlabel("k")
ax1.set_ylabel("wcss")

ax2 = ax1.twinx()
ax2.plot(k_range,scores[:len(k_range)],marker = "x",color = "blue")
ax2.set_ylabel("ss")
```

```python
kmeans = KMeans(n_clusters = 4,random_state = 42)
labels_kmeans = kmeans.fit_predict(x_pca)
```

```python
fig = plt.figure(figsize = (8,6))
ax = fig.add_subplot(111,projection = "3d")
ax.scatter(
    x_pca[:,0],x_pca[:,1],x_pca[:,2],c = labels_kmeans
)
```

```python
#agglomerative
```

```python
from sklearn.cluster import AgglomerativeClustering
```

```python
agg_clf = AgglomerativeClustering(n_clusters = 4,linkage = "ward")
```

```python
agg_labels = agg_clf.fit_predict(x_pca)
```

```python
fig = plt.figure(figsize = (8,6))
ax = fig.add_subplot(111,projection = "3d")
ax.scatter(
    x_pca[:,0],x_pca[:,1],x_pca[:,2],c = agg_labels
)
```
